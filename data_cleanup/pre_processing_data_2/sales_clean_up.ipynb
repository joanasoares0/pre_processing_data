```python
df = df.drop_duplicates()
df.shape
```

```python
# mssing values - numeric replace with median, categorical replace with Unkown

df = df.copy()

# numeric
for col in df.select_dtypes(include="number").columns:
    # if there's a col you don't want to apply
    # if col == 'col_name':
        #continue
    df[col] = df[col].fillna(df[col].median())
    
# categorical
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna('unknown')

print( df.isnull().sum()[df.isnull().sum()>0] )
```

```python
# standardize messy text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = (df[col]
               .str.strip()
               .str.lower()
               .str.replace(r'\s+', ' ', regex=True)  # collapse internal double spaces
               )

df.select_dtypes(include="object").apply(lambda s: s.unique())
```

```python
# mapping dicts to standardize messy category values (give a snippet to ai so it generates the dicts)
country_map = {
    'u.k.': 'united kingdom',
    'uk': 'united kingdom',
}

product_category_map = {
    'clothin': 'clothing',
    'cloths': 'clothing',
    'electronic': 'electronics',
    'electronicss': 'electronics',
}

payment_method_map = {
    'banktransfer': 'bank transfer',
}

df['country'] = df['country'].replace(country_map)
df['product_category'] = df['product_category'].replace(product_category_map)
df['payment_method'] = df['payment_method'].replace(payment_method_map)

#or
#df['country'] = df['country'].map(country_map).fillna(df['country'])
#df['product_category'] = df['product_category'].map(product_category_map).fillna(df['product_category'])
#df['payment_method'] = df['payment_method'].map(payment_method_map).fillna(df['payment_method'])

print(sorted(df['country'].unique()))
print(sorted(df['product_category'].unique()))
print(sorted(df['payment_method'].unique()))
print(df.head(10))
```

```python
# $ , ; .

#remove characteres in text 
df['price'] = (
    df['price']
    .astype(str)
    .str.replace( r'[$£,]', '', regex=True )
    .str.strip()
) 

# convert to correct dtype
df['price'] = pd.to_numeric(df['price'], errors='coerce')

df['price'].sample(10)
```

```python
# handling dates

date_cols = ['signup_date', 'order_date']
#or
# date_cols = [c for c in df.columns if 'date' in c]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], dayfirst=False, format='mixed', errors='coerce')
    
failed = df[['signup_date', 'order_date']].isnull().sum()
print(failed)
df.sample(10)
```

```python
# extract year, month, day
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day_of_week'] = df['order_date'].dt.day_name()

df[['order_date', 'order_year', 'order_month', 'order_day_of_week']]
```

```python
# apply logic tests
# flag rows that fail plausibility / consistency checks instead of silently altering data

df['age_valid'] = df['age'].between(0, 120)
df['rating_valid'] = df['rating'].between(1, 5)
df['price_valid'] = df['price'] >= 0
df['date_order_valid'] = df['order_date'] >= df['signup_date']
df['email_valid'] = df['email'].str.match(r'^[^@\s]+@[^@\s]+\.[^@\s]+$')

checks = ['age_valid', 'rating_valid', 'price_valid', 'date_order_valid', 'email_valid']
print(df[checks].sum())

print()

print("rows failing at least one check:", (~df[checks].all(axis=1)).sum())

df[~df[checks].all(axis=1)]
```

1. df[checks] — selects just the 5 boolean flag columns (age_valid, rating_valid, price_valid, date_order_valid, email_valid), giving a mini DataFrame of True/False values, one row per original row.

2. .all(axis=1) — for each row, checks whether all values across those columns are True. 
axis=1 means "collapse across columns, row-wise" (as opposed to axis=0, which would collapse down each column). 
Result: one boolean per row — True only if that row passed every check.

3. ~ — negates it, flipping True↔False. 
Now it's True for rows that failed at least one check (since "not all passed" = "at least one failed").

4. df[...] — boolean-indexes the full DataFrame with that mask, keeping only the rows that failed something.

So the whole line reads as: "give me every row where at least one of the five validity checks came back False."
